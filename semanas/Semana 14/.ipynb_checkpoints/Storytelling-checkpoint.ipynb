# Acto 1: Nivel estratégico

```python
import pyspark.sql.functions as F
# Calculamos la participación de cada marca para el Director General
total = df_completos.count()

kpi_estrategico = df_completos.groupBy("marca") \
    .agg(
        F.count("marca").alias("Cantidad de notebooks"),
        F.round((F.count("marca") / total) * 100, 4).alias("Participacion_Porcentaje")
    ).orderBy(F.desc("Participacion_Porcentaje"))

print(" [KPI ESTRATÉGICO] - ESTRUCTURA DE PORTAFOLIO POR PROVEEDOR")
kpi_estrategico.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pyspark.sql.functions as F

# 2. Conversión a Pandas solo para graficar
df_est_pandas = kpi_estrategico.toPandas()

# 3. Diseño del Gráfico Estratégico
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Creamos gráfico de barras horizontales
ax = sns.barplot(
    x="Participacion_Porcentaje", 
    y="marca", 
    data=df_est_pandas, 
    hue="marca",         # <-- Mapeo explícito de color
    palette="Oranges_r",
    legend=False         # <-- Ocultamos la leyenda redundante
)

# Añadimos las etiquetas de porcentaje al final de cada barra
for p in ax.patches:
    ax.annotate(f'{p.get_width():.5f}%', 
                (p.get_width() + 1, p.get_y() + p.get_height()/2), 
                va='center', ha='left', fontsize=11, fontweight='bold')

plt.title("VISTA ESTRATÉGICA: Concentración del Portafolio por Marca (Frecuencia: Mensual)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Participación en el Catálogo (% del Total notebooks)", fontsize=12)
plt.ylabel("Marca / Fabricante", fontsize=12)
plt.xlim(0, df_est_pandas["Participacion_Porcentaje"].max() + 10) # Espacio para el texto
plt.tight_layout()
plt.show()
```

# Acto 2: Nivel táctico

```python
# Analizamos la dispersión y bandas de precios para el Gerente Comercial
kpi_tactico = df_completos.groupBy("marca") \
    .agg(
        F.min("precio_numerico").alias("Precio_Minimo"),
        F.round(F.avg("precio_numerico"), 2).alias("Precio_Promedio"),
        F.max("precio_numerico").alias("Precio_Maximo"),
        F.round(F.stddev("precio_numerico"), 2).alias("Desviacion_Estandar_Precios")
    ) \
    .fillna(0, subset=["Desviacion_Estandar_Precios"]) \
    .orderBy(F.desc("Precio_Promedio"))

print(" [KPI TÁCTICO] - MATRIZ DE VOLATILIDAD Y COMPETITIVIDAD DE PRECIOS")
kpi_tactico.show()
```

```python
import matplotlib.pyplot as plt

# 1. Aseguramos que tenemos los datos en Pandas
df_tac_pandas = kpi_tactico.toPandas()

# 2. Configuración del gráfico
plt.figure(figsize=(14, 7))

# Líneas verticales que representan el rango (Min a Max)
plt.vlines(x=df_tac_pandas['marca'], 
           ymin=df_tac_pandas['Precio_Minimo'], 
           ymax=df_tac_pandas['Precio_Maximo'], 
           colors='gray', alpha=0.5, linewidth=2, label="Rango de Precios (Min a Max)")

# Puntos específicos
plt.scatter(df_tac_pandas['marca'], df_tac_pandas['Precio_Promedio'], color='navy', s=100, label="Precio Promedio", zorder=3)
plt.scatter(df_tac_pandas['marca'], df_tac_pandas['Precio_Minimo'], color='green', marker='^', s=60, label="Precio Mínimo", zorder=3)
plt.scatter(df_tac_pandas['marca'], df_tac_pandas['Precio_Maximo'], color='red', marker='v', s=60, label="Precio Máximo", zorder=3)

# Etiquetas y Estilo
plt.title("VISTA TÁCTICA: Análisis de Bandas de Precios por Marca", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Marcas", fontsize=12)
plt.ylabel("Precio ($)", fontsize=12) # Ajustado ya que no es necesariamente por Kg
plt.xticks(rotation=45, ha='right') # ha='right' ayuda a que los nombres no se corten
plt.legend(loc="upper left", frameon=True)
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

# Acto 3: Nivel operativo

```python
from pyspark.sql import functions as F
kpi_operacional_alertas = df_completos.filter(F.col("precio_numerico") < 150000) \
    .select("identificador", "tienda", "precio_numerico") \
    .orderBy("precio_numerico")

print(f" [ALERTA OPERATIVA]: PRODUCTOS DE BAJO COSTO (Total alertas: {kpi_operacional_alertas.count()})")
kpi_operacional_alertas.show(10)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df_alertas_pd = kpi_operacional_alertas.toPandas()
df_alertas_pd = df_alertas_pd.sort_values('precio_numerico', ascending=False)

fig, ax = plt.subplots(figsize=(10, 5))

# Barras horizontales
bars = ax.barh(df_alertas_pd['identificador'], df_alertas_pd['precio_numerico'], 
               color='#2E86AB', alpha=0.8, edgecolor='white')

# Colores según umbral
for i, (bar, precio) in enumerate(zip(bars, df_alertas_pd['precio_numerico'])):
    if precio < 120000:
        bar.set_color('#D9534F')  # Rojo → no ofertar
    elif precio < 140000:
        bar.set_color('#F0AD4E')  # Naranja → oferta media
    else:
        bar.set_color('#5BC0DE')  # Azul → oferta posible

# Etiquetas
ax.set_title("Clasificación de Productos por Precio - Estrategia de Ofertas", fontsize=12)
ax.set_xlabel("Precio ($)")
ax.set_xlim(80000, 160000)

# Leyenda personalizada
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#5BC0DE', label='Posible oferta (>$140k)'),
    Patch(facecolor='#F0AD4E', label='Oferta media ($120k-$140k)'),
    Patch(facecolor='#D9534F', label='No ofertar (<$120k)')
]
ax.legend(handles=legend_elements, loc='lower right')

plt.tight_layout()
plt.show()
```
