# Parte 1: Análisis con processed_data (sin clusters)

## Conexión a MongoDB y carga de processed_data

Objetivo: KPIs estratégicos, tácticos y operativos (concentración de mercado, precios, dispersión, outliers, canales).


```python
# Celda 1: Conexión y carga de processed_data
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import col, round, count, avg, min, max, stddev, expr
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Asegúrate de usar tu cadena de conexión real (con usuario y contraseña)
uri = "mongodb+srv://JavieraMorales:Javi28122004@cluster0.as1zz9d.mongodb.net/ecommerce.processed_data?retryWrites=true&w=majority"

spark = SparkSession.builder \
    .appName("KPIs_processed_data") \
    .config("spark.mongodb.read.connection.uri", uri) \
    .config("spark.jars.packages", "org.mongodb.spark:mongo-spark-connector_2.12:10.3.0") \
    .getOrCreate()

df = spark.read.format("mongodb") \
    .option("database", "ecommerce") \
    .option("collection", "processed_data") \
    .load()

print(f"Registros cargados: {df.count()}")
```

## KPI ESTRATÉGICO

###  Concentración del mercado por fabricante

```python
# Celda 2: KPI Estratégico – Concentración por marca
total = df.count()
kpi_marca = df.groupBy("marca").agg(
    count("marca").alias("Cantidad"),
    round((count("marca") / total) * 100, 2).alias("Participacion_%")
).orderBy(F.desc("Participacion_%"))

print("=== Concentración del mercado por fabricante ===")
kpi_marca.show(10)

df_plot = kpi_marca.toPandas()
plt.figure(figsize=(10,6))
sns.barplot(data=df_plot, x="Participacion_%", y="marca", hue="marca", palette="Blues_r", legend=False)
plt.title("Concentración del mercado por fabricante", fontweight="bold")
plt.xlabel("Participación (%)")
plt.ylabel("Marca")
plt.tight_layout()
plt.show()
```

### Precio promedio por marca

```python
# Celda 3: KPI Estratégico – Precio promedio por marca
kpi_precio = df.groupBy("marca").agg(
    round(avg("precio_numerico"), 0).alias("Precio_Promedio_CLP")
).orderBy(F.desc("Precio_Promedio_CLP"))

print("=== Precio promedio por marca ===")
kpi_precio.show(10)

df_precio = kpi_precio.toPandas()
plt.figure(figsize=(10,6))
sns.barplot(data=df_precio, x="Precio_Promedio_CLP", y="marca", hue="marca", palette="Reds_r", legend=False)
plt.title("Precio promedio por fabricante", fontweight="bold")
plt.xlabel("Precio promedio (CLP)")
plt.ylabel("Marca")
plt.tight_layout()
plt.show()
```

### Rango de precios por tienda (mínimo y máximo)

```python
# Celda 4: KPI Estratégico – Rango de precios por tienda (mínimo y máximo)
kpi_rango = df.groupBy("tienda").agg(
    round(min("precio_numerico"), 0).alias("Min"),
    round(max("precio_numerico"), 0).alias("Max")
).orderBy("tienda")

print("=== Rango de precios por tienda ===")
kpi_rango.show(10)

df_rango = kpi_rango.toPandas()
plt.figure(figsize=(10,5))
for _, row in df_rango.iterrows():
    plt.plot([row["Min"], row["Max"]], [row["tienda"], row["tienda"]], "o-", color="gray")
plt.scatter(df_rango["Min"], df_rango["tienda"], color="green", label="Mínimo", zorder=3)
plt.scatter(df_rango["Max"], df_rango["tienda"], color="red", label="Máximo", zorder=3)
plt.title("Rango de precios por tienda", fontweight="bold")
plt.xlabel("Precio (CLP)")
plt.ylabel("Tienda")
plt.legend()
plt.tight_layout()
plt.show()
```

## KPI TÁCTICO

### Dispersión de precios por tienda

```python
# Celda 5: KPI Táctico – Dispersión de precios por tienda
kpi_disp = df.groupBy("tienda").agg(
    round(avg("precio_numerico"), 0).alias("Promedio"),
    round(stddev("precio_numerico"), 0).alias("Desv_Estandar")
).orderBy(F.desc("Promedio"))

print("=== Dispersión de precios por tienda ===")
kpi_disp.show(10)

df_disp = kpi_disp.toPandas()
plt.figure(figsize=(10,5))
plt.errorbar(x=df_disp["tienda"], y=df_disp["Promedio"],
             yerr=df_disp["Desv_Estandar"], fmt="o", capsize=5, color="navy")
plt.title("Dispersión de precios por tienda (promedio ± desviación)", fontweight="bold")
plt.xlabel("Tienda")
plt.ylabel("Precio (CLP)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## KPI OPERACIONAL

### Concentración de canales (tiendas)

```python
# Celda 6: KPI Operativo – Concentración de canales (tiendas)
kpi_canales = df.groupBy("tienda").agg(
    count("tienda").alias("Cantidad"),
    round((count("tienda") / total) * 100, 2).alias("Participacion_%")
).orderBy(F.desc("Cantidad"))

print("=== Concentración de canales competitivos ===")
kpi_canales.show(10)

df_canales = kpi_canales.toPandas()
plt.figure(figsize=(10,5))
sns.barplot(data=df_canales, x="Participacion_%", y="tienda", hue="tienda", palette="Greens_r", legend=False)
plt.title("Participación de cada tienda en el catálogo", fontweight="bold")
plt.xlabel("Participación (%)")
plt.ylabel("Tienda")
plt.tight_layout()
plt.show()
```

### Alertas de outliers de precio por marca

```python
# Celda 7: KPI Operativo – Alertas de outliers de precio por marca (método IQR)
quantiles = df.groupBy("marca").agg(
    expr("percentile_approx(precio_numerico, 0.25)").alias("Q1"),
    expr("percentile_approx(precio_numerico, 0.75)").alias("Q3")
).withColumn("IQR", col("Q3") - col("Q1"))

df_con_iqr = df.join(quantiles, on="marca", how="left")
outliers = df_con_iqr.filter(
    (col("precio_numerico") > col("Q3") + 1.5 * col("IQR")) &
    col("Q1").isNotNull()
)

print(f"=== Alertas de outliers de precio (total: {outliers.count()}) ===")
outliers.select("marca", "precio_numerico", "ram_numero", "almacenamiento_numero", "tienda").show(10)

# Gráfico
df_all = df.select("marca", "precio_numerico", "ram_numero").toPandas()
outliers_pd = outliers.select("marca", "precio_numerico", "ram_numero").toPandas()

plt.figure(figsize=(10,6))
plt.scatter(df_all["ram_numero"], df_all["precio_numerico"], alpha=0.4, color="gray", label="Productos normales")
plt.scatter(outliers_pd["ram_numero"], outliers_pd["precio_numerico"], color="crimson", s=80, label="Outliers (precio elevado)", edgecolor="black")
plt.title("Detección de outliers de precio por marca", fontweight="bold")
plt.xlabel("RAM (GB)")
plt.ylabel("Precio (CLP)")
plt.legend()
plt.tight_layout()
plt.show()
```

## Finalizar sesión Spark (para poder reiniciar el kernel limpio)

```python
# Celda 8: Cerrar sesión Spark antes de reiniciar el kernel
spark.stop()
print("Sesión Spark finalizada. Ahora puedes reiniciar el kernel (Kernel → Restart).")
```

# Parte 2: Análisis con notebooks_con_clusters (segmentación)

## Conexión a MongoDB y carga de notebooks_con_clusters

```python
# Celda 1: Conexión y carga de notebooks_con_clusters
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

uri_clusters = "mongodb+srv://JavieraMorales:Javi28122004@cluster0.as1zz9d.mongodb.net/ecommerce.notebooks_con_clusters?retryWrites=true&w=majority"

spark = SparkSession.builder \
    .appName("KPIs_clusters") \
    .config("spark.mongodb.read.connection.uri", uri_clusters) \
    .config("spark.jars.packages", "org.mongodb.spark:mongo-spark-connector_2.12:10.3.0") \
    .getOrCreate()

df_clust = spark.read.format("mongodb") \
    .option("database", "ecommerce") \
    .option("collection", "notebooks_con_clusters") \
    .load()

print(f"Registros cargados: {df_clust.count()}")
```

## Resumen de clusters (K-Means)

```python
# Celda 2: Resumen de clusters (K-Means)
total = df_clust.count()
resumen_kmeans = df_clust.groupBy("prediction").agg(
    F.count("*").alias("Cantidad"),
    F.round((F.count("*") / total) * 100, 2).alias("Porcentaje"),
    F.round(F.avg("precio_numerico"), 0).alias("Precio_Promedio"),
    F.round(F.avg("ram_numero"), 1).alias("RAM_Promedio"),
    F.round(F.avg("almacenamiento_numero"), 0).alias("Almacenamiento_Promedio")
).orderBy("prediction")

print("=== Segmentación de gamas (K-Means) ===")
resumen_kmeans.show()

pdf = df_clust.select("ram_numero", "precio_numerico", "prediction").toPandas()
plt.figure(figsize=(10,6))
sns.scatterplot(data=pdf, x="ram_numero", y="precio_numerico", hue="prediction", palette="Set2", alpha=0.6)
plt.title("Segmentación por cluster (RAM vs Precio)", fontweight="bold")
plt.xlabel("RAM (GB)")
plt.ylabel("Precio (CLP)")
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()
```

## Resumen de DBSCAN

```python
# Celda 3: Resumen de DBSCAN
resumen_dbscan = df_clust.groupBy("dbscan_cluster").agg(
    F.count("*").alias("Cantidad"),
    F.round((F.count("*") / total) * 100, 2).alias("Porcentaje")
).orderBy("dbscan_cluster")

print("=== Clustering DBSCAN ===")
resumen_dbscan.show()

ruido = df_clust.filter(F.col("dbscan_cluster") == -1)
print(f"Puntos considerados ruido (outliers por densidad): {ruido.count()}")
ruido.select("marca_especif", "precio_numerico", "ram_numero", "almacenamiento_numero").show(5)
```

## Visualización DBSCAN (RAM vs Precio)

```python
# Celda 4: Visualización DBSCAN (RAM vs Precio)
pdf_db = df_clust.select("ram_numero", "precio_numerico", "dbscan_cluster").toPandas()
plt.figure(figsize=(10,6))
sns.scatterplot(data=pdf_db, x="ram_numero", y="precio_numerico", hue="dbscan_cluster", palette="Set1", alpha=0.6)
plt.title("Clustering DBSCAN (RAM vs Precio)", fontweight="bold")
plt.xlabel("RAM (GB)")
plt.ylabel("Precio (CLP)")
plt.legend(title="DBSCAN cluster")
plt.tight_layout()
plt.show()
```

## Celda 5: Cerrar sesión

```python
# Celda 5: Cerrar sesión
spark.stop()
print("Análisis de clusters completado.")
```

# Parte 3: Creación del CSV completo


```python
# exportar_csv.py
from pyspark.sql import SparkSession
from pyspark.sql.functions import row_number, col, desc
from pyspark.sql.window import Window
import os

# ===== CONFIGURACIÓN (CAMBIA AQUÍ TUS CREDENCIALES) =====
MONGO_USER = "JavieraMorales"
MONGO_PWD  = "Javi28122004"
MONGO_CLUSTER = "cluster0.as1zz9d.mongodb.net"
DATABASE = "ecommerce"
# ========================================================

URI_BASE = f"mongodb+srv://{MONGO_USER}:{MONGO_PWD}@{MONGO_CLUSTER}/{DATABASE}."

spark = SparkSession.builder \
    .appName("Exportar_CSV_Limpio") \
    .config("spark.mongodb.read.connection.uri", URI_BASE + "processed_data") \
    .config("spark.jars.packages", "org.mongodb.spark:mongo-spark-connector_2.12:10.3.0") \
    .getOrCreate()

# Cargar colecciones
df_proc = spark.read.format("mongodb") \
    .option("database", DATABASE) \
    .option("collection", "processed_data") \
    .load()

df_clust = spark.read.format("mongodb") \
    .option("database", DATABASE) \
    .option("collection", "notebooks_con_clusters") \
    .load()

# Deduplicar clusters: quedarse con el registro más reciente por producto
# (usamos fecha_clustering_dbscan como timestamp; si no existe, usamos otra columna o el orden natural)
window_spec = Window.partitionBy("precio_numerico", "ram_numero", "almacenamiento_numero") \
                    .orderBy(desc("fecha_clustering_dbscan"))

df_clust_dedup = df_clust.withColumn("rn", row_number().over(window_spec)) \
                         .filter(col("rn") == 1) \
                         .drop("rn")

# Left join
df_join = df_proc.join(df_clust_dedup,
                       on=["precio_numerico", "ram_numero", "almacenamiento_numero"],
                       how="left")

# Guardar CSV
output_dir = "/home/jovyan/work/semanas/Semana 15"
os.makedirs(output_dir, exist_ok=True)
csv_path = os.path.join(output_dir, "datos_notebooks_dashboard.csv")

df_join.toPandas().to_csv(csv_path, index=False)
print(f"CSV guardado en: {csv_path}")
print(f"Registros: {df_join.count()}, columnas: {len(df_join.columns)}")

spark.stop()
```
