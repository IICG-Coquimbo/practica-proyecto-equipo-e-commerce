¿El problema es el algoritmo o los datos?
El problema principal reside en los datos. El algoritmo K-Means es una herramienta matemática eficiente que cumple su función: agrupar puntos por cercanía en un espacio n-dimensional. Sin embargo,  si los datos de entrada están sesgados o incompletos, los clústeres resultantes no tendrán sentido lógico de negocio. 

¿Cómo lo solucionaría en el mundo real? (Sin hacer trampa)

Costos Logísticos: Si el producto es importado o nacional, ya que los aranceles y fletes determinan el precio base.
Margen de Ganancia: El "markup" de cada tienda; algunas cadenas tienen costos operativos más altos que las tiendas 100% online.
Antigüedad del Modelo (Obsolescencia): Un notebook puede tener mucha RAM, pero si el procesador es de hace 4 generaciones, su precio caerá. Este dato de "generación" no siempre es fácil de scrapear.
Valor de Marca: El prestigio de la marca es un factor subjetivo que el algoritmo no entiende como "tecnología", pero que el mercado paga.
