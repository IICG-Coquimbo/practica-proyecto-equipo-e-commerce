##1. ¿El problema es el algoritmo o los datos?

Yo creo que el problema son los datos. Un algoritmo aprende de lo que le pones si los datos están sucios, desordenados o tienen cosas que no deberían estar como productos que no son computadores o precios en cero, por más que el algoritmo intente hacer magia, no va a funcionar bien. Es como intentar cocinar una receta con ingredientes podridos por buena que sea la receta, el plato saldrá mal. En este caso, los datos tenían duplicados, textos basura, valores vacíos y hasta productos de motos o repuestos. Con esa base, ningún algoritmo puede dar resultados confiables.

##2. ¿Cómo lo solucionarían en el mundo real sin hacer trampa? ¿Qué datos faltan?


En la vida real, empezaría por categorizar bien los datos desde el principio: separar lo que es un computador de lo que no, ordenar cada característica (RAM, almacenamiento, procesador) en su propia columna, y limpiar todo lo que esté mal escrito o incompleto. Para productos veterinarios, los datos que faltan y que de verdad influyen en el precio son cosas como el costo de la materia prima, si el producto es importado o nacional, el peso neto, la fecha de vencimiento y el margen de ganancia que pone cada tienda. Sin esos datos, cualquier predicción de precio va a estar coja. La solución no es trampear inventando números, sino exigir que la captura de datos sea limpia y completa desde el origen.
