# Respuestas:

### 1. ¿El problema es el algoritmo o los datos?).

El problema son los datos, no el algoritmo. K-means clusterizó bien los notebooks por precio, RAM y almacenamiento, pero predecir el precio solo con esas variables falla. La dispersión es enorme: notebooks con 16 GB RAM/512 GB SSD van desde $170 000 hasta $6 000 000. Faltan variables clave como procesador, gráfica, pantalla, peso, etc.

### 2. ¿Cómo lo solucionarían sin hacer trampa? ¿Qué datos faltan?

Sin inventar datos, se debe enriquecer el scraping capturando: modelo exacto del procesador, tarjeta gráfica, tipo de pantalla, peso, condición (nuevo/reacondicionado), garantía. Con esas variables, un Random Forest o regresión múltiple darían predicciones útiles. Con mis datos actuales, el clustering funciona bien, pero la regresión sería muy imprecisa (R² bajo, error > $500 000).
